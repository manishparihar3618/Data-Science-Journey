```python
import numpy as np
import pandas as pd
from sklearn.datasets import make_moons

X, y = make_moons(
    n_samples=1000,
    noise=0.20,
    random_state=42
)

df = pd.DataFrame(X, columns=["feature_1", "feature_2"])
df["target"] = y

df.to_csv("svm_practice.csv", index=False)

print(df.head())
print(df.shape)
print(df["target"].value_counts())
```
